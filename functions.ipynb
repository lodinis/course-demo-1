```python
import numpy as np
import pandas as pd
```

```python
def generate_random_list(a: int, b: int, length: int):
    arr = np.random.randint(a, b + 1, size=length)

    return arr.tolist()
```

```python
def build_frequency_dict(data: list):
    # value_counts() считает частоты
    return pd.Series(data).value_counts().to_dict()
```

```python
def calculate_statistics(data: list):
    arr = np.array(data)
    # Поиск моды через np.unique:
    vals, counts = np.unique(arr, return_counts=True)
    max_count = np.max(counts)
    modes = vals[counts == max_count].tolist()

    return {
        "min": float(np.min(arr)),
        "mean": float(np.mean(arr)),
        "median": float(np.median(arr)),
        "mode": modes,
        "max": float(np.max(arr)),
    }
```

```python
# Генерация случайного списка
a = 1
b = 5
length = 20

sample = generate_random_list(a, b, length)
print(f"Сгенерированный список: {sample}\n")
```

```python
# Проверка частотного ряда
print(f"Частотный ряд для выборки: {build_frequency_dict(sample)}\n")
```

```python
print(f"Стат данные для выборки: {calculate_statistics(sample)}\n")
```
